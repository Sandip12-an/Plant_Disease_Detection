# 01. Exploratory Data Analysis (EDA) — Plant Disease Detection

This notebook explores the class distribution, image dimensions, color spaces, and visual characteristics of the PlantVillage and PlantDoc datasets.

```python
import os
import glob
import json
import matplotlib.pyplot as plt
from PIL import Image
from collections import Counter

DATA_DIR = "../../data/raw"
print("Inspecting raw dataset directory:", os.path.abspath(DATA_DIR))
```

```python
with open("../model_registry/class_labels.json", "r") as f:
    class_labels = json.load(f)

print(f"Total target classes: {len(class_labels)}")
crops = sorted(list(set([c.split('___')[0] for c in class_labels])))
print(f"Total crop species ({len(crops)}):", ", ".join(crops))
```

```python
# Count images per class
class_counts = {}
for cls in class_labels:
    folder = os.path.join(DATA_DIR, cls)
    count = len(glob.glob(os.path.join(folder, "*.*"))) if os.path.exists(folder) else 0
    class_counts[cls] = count

# Display class distribution bar chart
plt.figure(figsize=(12, 6))
plt.barh(list(class_counts.keys())[:15], list(class_counts.values())[:15], color='forestgreen')
plt.xlabel("Image Count")
plt.title("Class Distribution (Top 15 Classes)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
