# 03. Transfer Learning Experiments (MobileNetV2 vs. EfficientNet-B0 vs. ResNet50)

This notebook provides systematic comparisons across pre-trained feature extractors, evaluating trade-offs between inference speed, parameter footprint, and validation accuracy on plant disease classification.

```python
import torchvision.models as models
import torch

architectures = {
    "MobileNetV2": models.mobilenet_v2(weights=None),
    "EfficientNet-B0": models.efficientnet_b0(weights=None),
    "ResNet50": models.resnet50(weights=None)
}

print(f"{'Architecture':<20} | {'Total Parameters':<18} | {'Est. Size (MB)':<15}")
print("-" * 60)
for name, net in architectures.items():
    params = sum(p.numel() for p in net.parameters())
    size_mb = (params * 4) / (1024 * 1024)
    print(f"{name:<20} | {params:<18,} | {size_mb:<15.2f}")
```
